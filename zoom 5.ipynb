# Studi Kasus: Perancangan Dasbor Analitik Penjualan Berbasis Prinsip IMK dan Visualisasi Data

## 1. Latar Belakang & Skenario Masalah

Sebuah perusahaan ritel multinasional menghadapi masalah penurunan kepuasan pengguna (*user satisfaction*) terhadap dasbor internal yang digunakan oleh para manajer wilayah. Berdasarkan evaluasi awal dengan prinsip **User-Centered Design (UCD)** dan **Heuristic Evaluation**:

* **Masalah (*Pain Points*)**: Dasbor lama terlalu padat dengan angka mentah, menggunakan warna yang tidak konsisten, memiliki banyak *chartjunk* (hiasan visual berlebihan seperti efek 3D dan garis grid tebal), serta membebani memori kognitif pengguna (*high cognitive load*).
* **Solusi Berbasis IMK**: Mengubah tata letak dasbor dengan menerapkan prinsip Gestalt (mengelompokkan informasi terkait secara berdekatan), membatasi beban kognitif (*extraneous load*), serta menggunakan atribut visual preattentive (warna kontras untuk anomali atau data penting). Sesuai prinsip **"Overview first, zoom and filter, details on demand"**, dasbor dirancang agar manajer dapat melihat ringkasan performa utama dalam waktu kurang dari 5 detik.

---

## 2. Kebutuhan Sistem (Functional & Non-Functional)

* **Kebutuhan Fungsional**:
1. Sistem dapat menampilkan ringkasan metrik utama (*Total Penjualan, Rata-rata Transaksi, dan Pertumbuhan*).
2. Sistem menyediakan grafik tren penjualan bulanan dan perbandingan kategori produk.
3. Sistem memiliki fitur filter interaktif berdasarkan wilayah (*region*).


* **Kebutuhan Non-Fungsional**:
1. Waktu muat (*load time*) visualisasi di bawah 2 detik.
2. Antarmuka bersih dari elemen dekoratif yang tidak perlu (*high data-ink ratio*).



---

## 3. Implementasi Kode Python untuk Analisis & Visualisasi Dasbor

Berikut adalah kode Python menggunakan `pandas` dan `matplotlib` untuk menghasilkan visualisasi data yang bersih, minimalis, dan sesuai dengan standar psikologi kognitif visual (mengurangi *chartjunk*):

```
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Bulan': ['Jan', 'Feb', 'Mar', 'Apr', 'Mei', 'Jun', 'Jul', 'Agu', 'Sep', 'Okt', 'Nov', 'Des'],
    'Wilayah_A': [120, 135, 150, 142, 160, 175, 190, 185, 210, 225, 240, 280],
    'Wilayah_B': [90, 105, 110, 125, 130, 145, 160, 155, 170, 190, 205, 240]
}

df = pd.DataFrame(data)

print("--- Ringkasan Data Performa Penjualan (Data-Ink Ratio Optimal) ---")
print(df.to_string(index=False))

total_a = df['Wilayah_A'].sum()
total_b = df['Wilayah_B'].sum()
print(f"\n[Ringkasan Eksekutif]")
print(f"Total Penjualan Wilayah A: Rp {total_a} Juta")
print(f"Total Penjualan Wilayah B: Rp {total_b} Juta")

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(df['Bulan'], df['Wilayah_A'], label='Wilayah A (Prioritas Utama)', color='#1f77b4', marker='o', linewidth=2.5)
ax.plot(df['Bulan'], df['Wilayah_B'], label='Wilayah B', color='#ff7f0e', marker='s', linewidth=2, linestyle='--')

ax.set_title('Analisis Tren Kinerja Penjualan Bulanan (2026)', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Bulan', fontsize=11, fontweight='semibold')
ax.set_ylabel('Penjualan (Dalam Juta IDR)', fontsize=11, fontweight='semibold')

ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(frameon=True, facecolor='white', edgecolor='none')

plt.tight_layout()

plt.show()
```

## 4. Evaluasi Penggunaan (*Usability Testing*)

Setelah implementasi dasbor berbasis kode di atas, dilakukan pengujian menggunakan **10 Hebat Heuristik Nielsen** dan *Usability Testing* kepada pengguna nyata (para manajer):

1. **Visibility of System Status**: Menambahkan indikator status pemuatan data secara real-time.
2. **Match Between System and Real World**: Menggunakan istilah bisnis yang akurat dan familiar (*revenue, growth, regional filter*).
3. **Aesthetic and Minimalist Design**: Menghapus elemen 3D dan warna mencolok yang tidak perlu, sehingga fokus pengguna langsung tertuju pada tren grafik utama (*preattentive visual processing*).
